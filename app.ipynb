# ChatWell FastAPI notebook launcher

The application backend lives in `app.py`. The original Flask route code and the inline dependency installer have been replaced by FastAPI and the project dependency list in `pyproject.toml`. Run this notebook with the project `.venv` kernel from the repository root. The chatbot retrieves context locally and uses Groq's hosted `openai/gpt-oss-120b` model by default to generate answers.


## Install project dependencies

Use uv to create or synchronize `.venv`. Run `uv sync --locked` in a terminal before launching Jupyter; the cell below performs the same synchronization if needed. Register the notebook kernel with `uv run --locked python -m ipykernel install --prefix .venv --name chatwell-venv --display-name "ChatWell (.venv, Python 3.14.2)"`, then select that kernel in Jupyter.

## Run the application

Create a free Groq API key at <https://console.groq.com/keys> and place it in the project-root `.env` file as `GROQ_API_KEY=gsk_...`. Copy `.env.example` to `.env` if needed. The backend loads this file when chat starts; keep the key out of notebooks and Git. The server listens at <http://localhost:5000/>. Stop this cell to stop the server.


```python
import uvicorn

config = uvicorn.Config("app:app", host="localhost", port=5000)
server = uvicorn.Server(config)
await server.serve()
```
